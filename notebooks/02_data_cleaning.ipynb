```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
orders_dataset_file_path = "../data/raw/archive/olist_orders_dataset.csv"
order_items_dataset_file_path = "../data/raw/archive/olist_order_items_dataset.csv"
order_reviews_dataset_file_path = "../data/raw/archive/olist_order_reviews_dataset.csv"
```

```python
def read_dataset(dataset_path):
    dataset = pd.read_csv(dataset_path)
    return dataset

def convert_to_datetime(dataset, columns):
    for column in columns:
        dataset[column] = pd.to_datetime(dataset[column])
    return dataset
```

```python
original_orders_dataset = read_dataset(orders_dataset_file_path)
original_orders_items_dataset = read_dataset(order_items_dataset_file_path)
original_order_reviews_dataset = read_dataset(order_reviews_dataset_file_path)

display(original_orders_dataset.head(1))
display(original_orders_items_dataset.head(1))
display(original_order_reviews_dataset.head(1))
```

```python
orders_dataset = read_dataset(orders_dataset_file_path)
display(orders_dataset.dtypes)

orders_dataset.head(1)
orders_dataset_columns_to_datetime = [
    "order_purchase_timestamp", 
    "order_approved_at", 
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date" 
    ]
```

```python
order_items_dataset = read_dataset(order_items_dataset_file_path)

display(order_items_dataset.dtypes)
display(order_items_dataset.head(1))

order_items_dataset_columns_to_datetime = [
    "shipping_limit_date" 
    ]
```

```python
order_reviews_dataset = read_dataset(order_reviews_dataset_file_path)

display(order_reviews_dataset.dtypes)
display(order_reviews_dataset.head(1))

order_reviews_dataset_columns_to_datetime = [
    "review_creation_date",
    "review_answer_timestamp" 
    ]
```

Convert dates into datetimes format

```python
orders_dataset = convert_to_datetime(orders_dataset, orders_dataset_columns_to_datetime)
display(orders_dataset.dtypes)
orders_dataset.head(1)
```

```python
order_items_dataset = convert_to_datetime(order_items_dataset, order_items_dataset_columns_to_datetime)
display(order_items_dataset.dtypes)
order_items_dataset.head(1)
```

```python
order_reviews_dataset = convert_to_datetime(order_reviews_dataset, order_reviews_dataset_columns_to_datetime)
display(order_reviews_dataset.dtypes)
order_reviews_dataset.head(1)
```
