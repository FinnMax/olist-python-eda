```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
orders_dataset_file_path = "../data/raw/archive/olist_orders_dataset.csv"
order_items_dataset_file_path = "../data/raw/archive/olist_order_items_dataset.csv"
order_reviews_dataset_file_path = "../data/raw/archive/olist_order_reviews_dataset.csv"
geolocation_dataset_file_path = "../data/raw/archive/olist_geolocation_dataset.csv"
products_dataset_filepath = "../data/raw/archive/olist_products_dataset.csv"
```

```python
def read_dataset(dataset_path):
    dataset = pd.read_csv(dataset_path)
    return dataset

def convert_to_datetime(dataset, columns):
    for column in columns:
        dataset[column] = pd.to_datetime(dataset[column])
    return dataset
```

```python
original_orders_dataset = read_dataset(orders_dataset_file_path)
original_orders_items_dataset = read_dataset(order_items_dataset_file_path)
original_order_reviews_dataset = read_dataset(order_reviews_dataset_file_path)
original_geolocation_dataset = read_dataset(geolocation_dataset_file_path)
original_products_dataset = read_dataset(products_dataset_filepath)
```

Convert columns to datetime

```python
orders_dataset = read_dataset(orders_dataset_file_path)
display(orders_dataset.dtypes)

orders_dataset.head(1)
orders_dataset_columns_to_datetime = [
    "order_purchase_timestamp", 
    "order_approved_at", 
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date" 
    ]
```

```python
order_items_dataset = read_dataset(order_items_dataset_file_path)

display(order_items_dataset.dtypes)
display(order_items_dataset.head(1))

order_items_dataset_columns_to_datetime = [
    "shipping_limit_date" 
    ]
```

```python
order_reviews_dataset = read_dataset(order_reviews_dataset_file_path)

display(order_reviews_dataset.dtypes)
display(order_reviews_dataset.head(1))

order_reviews_dataset_columns_to_datetime = [
    "review_creation_date",
    "review_answer_timestamp" 
    ]
```

Convert dates into datetimes format

```python
orders_dataset = convert_to_datetime(orders_dataset, orders_dataset_columns_to_datetime)
display(orders_dataset.dtypes)
orders_dataset.head(1)
```

```python
order_items_dataset = convert_to_datetime(order_items_dataset, order_items_dataset_columns_to_datetime)
display(order_items_dataset.dtypes)
order_items_dataset.head(1)
```

```python
order_reviews_dataset = convert_to_datetime(order_reviews_dataset, order_reviews_dataset_columns_to_datetime)
display(order_reviews_dataset.dtypes)
order_reviews_dataset.head(1)
```

Missing values cleaning

```python
#Missing values in olist_orders_dataset
orders_dataset.head(1)
orders_dataset.dtypes

display(orders_dataset.isna().sum())

#order_approved_at check
approved_at_na = orders_dataset[orders_dataset["order_approved_at"].isna()]
approved_at_na["order_status"].value_counts()
approved_at_na[approved_at_na["order_status"] == "delivered"] #all the information is present except the order_approved_at
approved_at_na[approved_at_na["order_status"] == "created"] #most of the date columnsa re missing, except the estimated delivery date, suggesting that the system estimates it automatically

order_delivered_carrier_date_missing = orders_dataset[orders_dataset["order_delivered_carrier_date"].isna()] # most of the order status is not delvered so makes sense no date
order_delivered_carrier_date_missing["order_status"].value_counts()
order_delivered_carrier_date_missing[order_delivered_carrier_date_missing["order_status"] == "delivered"] #probably just missing timestamps no need to delete

order_delivered_customer_date = orders_dataset[orders_dataset["order_delivered_customer_date"].isna()]
order_delivered_customer_date["order_status"].value_counts() #most of the orders statsu is not delivered yet, also probably just missing some timestamps
# Based on this we can see that most of the orders which are missing the order_approved_at values are cancelled, with some exceptions which can suggest human or machine error.
# Delivered order status only have order_approved_at is missing, suggesting data anomalies. I think we can keep these rows
# Created orders also dont need to be removed becasue it makes sense for them to have missing values
```
